### `groupby()`
Conceptually, `groupby()` performs **split → apply → combine**:

1. split rows into groups;
2. apply a calculation to each group;
3. combine the group results.

Here we ask: *What is the mean elevation for each county?*

````python
county_elevation = (
    indiana.groupby("COUNTY_NAME")["ELEV_IN_M"]
           .mean()
)

county_elevation.head()
````

### Predict first
What does each row in `county_elevation` represent?

Now predict what changing `.mean()` to `.max()` would mean scientifically.

Each row of the `county_elevation` data shows the average elevation of recorded features within an Indiana county, but also excluding any missing values. If we replace `.mean()` with `.max()`, it will show the highest recorded elevation for each county instead. However, this does not mean necessarily that it is the actual highest point in that county because the dataset may not include every location within it.

````python
indiana.groupby("COUNTY_NAME")["ELEV_IN_M"].max().head()
````

### `agg()` — several summaries at once
`agg()` lets us apply multiple summary functions to each group.

````python
feature_elevation = (
    indiana.groupby("FEATURE_CLASS")["ELEV_IN_M"]
           .agg(["count", "mean", "median", "min", "max"])
)

feature_elevation.head(10)
````

## 7. Dates in Pandas — about 10 minutes

`DATE_CREATED` was initially read as text. To perform date-aware operations, convert it with `pd.to_datetime()`.

`errors="coerce"` means values that cannot be interpreted as dates become `NaT` (Not-a-Time) rather than stopping the program.

````python
work["DATE_CREATED"] = pd.to_datetime(work["DATE_CREATED"], errors="coerce")
work["DATE_EDITED"] = pd.to_datetime(work["DATE_EDITED"], errors="coerce")

work[["DATE_CREATED", "DATE_EDITED"]].dtypes
````

### FINAL CHALLENGE — Question 6: Your own question

**My question:** Among Indiana counties with at least 25 recorded stream elevations, which counties have the highest mean elevation of their named streams?

**Why this question is useful/interesting:** I beleive my question is interesting because it explores the geographic variation in stream elevations without allowing a county with only one or two stream observations to completely dominate the ranking. In addition, the minimum sample size rule makes the comparison more meaningful. Although the counts will still vary among the counties.

**Pandas operations I expect to need:** I need to first filter the dataset to include only streams, then group them by county to calculate the number of streams and their average elevation. Next, I only keep counties with at least 25 recorded streams to make the comparison more reliable. Then finally, I will sort the counties from highest to lowest average elevation and select the top five.

````python
county_stream_ranking = (
    indiana.loc[indiana["FEATURE_CLASS"] == "Stream"]
           .groupby("COUNTY_NAME")["ELEV_IN_M"]
           .agg(stream_count="count", mean_stream_elev_m="mean")
           .loc[lambda result: result["stream_count"] >= 25]
           .sort_values("mean_stream_elev_m", ascending=False)
           .head(5)
)
county_stream_ranking.round(2)
````

**Interpretation (2–3 sentences):** Randolph County has the highest average recorded stream elevation among the counties meeting the 25 observation threshold, with a value of 321.36 m across 78 streams, this is followed by Wayne County at 305.71 m and Henry County at 304.33 m. I believe this is consistent with the relatively high elevations of the named streams sampled within those counties, but to note the observations are not a random or area weighted sample. So it would be scientifically insensible to interpret these averages as each county's overall average elevation or as a direct measure of the streams elevation.
